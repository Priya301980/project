```python
import csv, math, os, random
from datetime import datetime, timedelta

rng = random.Random(42)
RAW, POL = os.path.join("data", "raw"), os.path.join("data", "policies")
os.makedirs(RAW, exist_ok=True); os.makedirs(POL, exist_ok=True)

CITIES = ["Chennai", "Mumbai", "Bengaluru", "Delhi", "Hyderabad", "Pune"]
PRODUCTS = [  # (name, category, price_inr)
    ("Basmati Rice 1kg", "Grocery", 129), ("Toor Dal 1kg", "Grocery", 165), ("Atta 5kg", "Grocery", 245),
    ("Sunflower Oil 1L", "Grocery", 155), ("Sugar 1kg", "Grocery", 48), ("Salt 1kg", "Grocery", 24),
    ("Tea Powder 250g", "Grocery", 120), ("Poha 500g", "Grocery", 42),
    ("Toned Milk 500ml", "Dairy", 29), ("Curd 400g", "Dairy", 45), ("Paneer 200g", "Dairy", 95),
    ("Butter 100g", "Dairy", 58), ("Cheese Slices 200g", "Dairy", 130), ("Ghee 500ml", "Dairy", 340),
    ("Eggs 12 Pack", "Dairy", 84), ("Lassi 200ml", "Dairy", 25),
    ("Potato Chips 52g", "Snacks", 20), ("Namkeen 200g", "Snacks", 60), ("Biscuits 150g", "Snacks", 30),
    ("Dark Chocolate 80g", "Snacks", 150), ("Instant Noodles 4 Pack", "Snacks", 56), ("Popcorn 100g", "Snacks", 45),
    ("Peanut Butter 340g", "Snacks", 210), ("Cold Coffee Can 180ml", "Snacks", 90),
    ("Shampoo 340ml", "Personal Care", 280), ("Face Wash 100ml", "Personal Care", 190),
    ("Toothpaste 150g", "Personal Care", 98), ("Soap 4 Pack", "Personal Care", 150),
    ("Hand Wash 250ml", "Personal Care", 99), ("Body Lotion 400ml", "Personal Care", 350),
    ("Deodorant 150ml", "Personal Care", 220), ("Sunscreen 50g", "Personal Care", 399),
    ("Wireless Earbuds", "Electronics", 1499), ("Power Bank 10000mAh", "Electronics", 1299),
    ("USB-C Cable 1m", "Electronics", 299), ("Phone Charger 20W", "Electronics", 799),
    ("Smart Bulb", "Electronics", 599), ("Bluetooth Speaker", "Electronics", 1999),
    ("Phone Stand", "Electronics", 349), ("Memory Card 64GB", "Electronics", 549)]
CAT_IDX = {}
for i, (_, c, _) in enumerate(PRODUCTS):
    CAT_IDX.setdefault(c, []).append(i)

def write(name, header, rows):
    with open(os.path.join(RAW, name), "w", newline="", encoding="utf-8") as f:
        w = csv.writer(f); w.writerow(header); w.writerows(rows)

def poisson(lam):
    L, k, p = math.exp(-lam), 0, 1.0
    while True:
        p *= rng.random()
        if p <= L:
            return k
        k += 1

def mangle_city(c):
    r = rng.random()
    return c.lower() if r < .20 else c.upper() if r < .32 else (" " + c + " ") if r < .42 else c

def mangle_gender(g):
    opts = {"male": ["male", "Male", "M", "m", "MALE"], "female": ["female", "Female", "F", "f", "FEMALE"]}[g]
    return rng.choice(opts)

# ---------------- products ----------------
write("products.csv", ["product_id", "product_name", "category", "price_inr"],
      [[f"P{i+1:03d}", n, c, p] for i, (n, c, p) in enumerate(PRODUCTS)])

# ---------------- customers ----------------
N_CUST = 1200
IMPOSSIBLE_AGE = {10: 280, 110: -4, 210: 0, 310: 150, 410: 999, 510: 200}
cust, arche = [], {}
for i in range(N_CUST):
    cid = f"C{i+1:04d}"
    r = rng.random()
    arche[cid] = "premium" if r < .20 else "frequent" if r < .60 else "rare"
    signup = datetime(2024, 1, 1) + timedelta(days=rng.randint(0, 455))
    age = IMPOSSIBLE_AGE.get(i, rng.randint(18, 60))
    age = "" if (i not in IMPOSSIBLE_AGE and rng.random() < .05) else age
    gender = "" if rng.random() < .03 else mangle_gender(rng.choice(["male", "female"]))
    cust.append([cid, mangle_city(rng.choice(CITIES)), signup.strftime("%Y-%m-%d"), age, gender])
write("customers.csv", ["customer_id", "city", "signup_date", "age", "gender"], cust)

# ---------------- orders ----------------
WIN_START, WIN_DAYS = datetime(2025, 4, 1), 183          # 2025-04-01 .. 2025-09-30
RATE = {"premium": 1.0, "frequent": 3.5, "rare": 0.5}   # expected orders per month
def pick_product(a):
    if a == "premium":
        pool = CAT_IDX["Electronics"] if rng.random() < .7 else CAT_IDX["Personal Care"] + CAT_IDX["Snacks"]
        return rng.choice(pool), rng.randint(1, 2)
    if a == "frequent":
        return rng.choice(CAT_IDX["Grocery"] + CAT_IDX["Dairy"] + CAT_IDX["Snacks"]), rng.randint(1, 3)
    return rng.choice(CAT_IDX["Grocery"] + CAT_IDX["Dairy"] + CAT_IDX["Snacks"] + CAT_IDX["Personal Care"]), rng.randint(1, 2)

rows, oid = [], 100000
for cid, a in arche.items():
    for _ in range(poisson(RATE[a] * 6.0)):
        oid += 1
        pi, qty = pick_product(a)
        ts = WIN_START + timedelta(days=rng.randint(0, WIN_DAYS - 1), hours=rng.randint(7, 22),
                                   minutes=rng.randint(0, 59), seconds=rng.randint(0, 59))
        disc = rng.choice([0, 0, 0, 5, 10, 15, 20, 25, 40])
        status = "cancelled" if rng.random() < .10 else "completed"
        amount = round(PRODUCTS[pi][2] * qty * (1 - disc / 100), 2)
        if status == "cancelled":
            deliv = ""
        else:
            deliv = rng.randint(46, 75) if rng.random() < .10 else max(8, min(45, int(round(rng.gauss(24, 8)))))
            if rng.random() < .06:
                deliv = ""
        rows.append({"order_id": f"ORD-{oid}", "customer_id": cid, "product_id": f"P{pi+1:03d}", "quantity": qty,
                     "order_ts": ts, "status": status, "discount_pct": disc, "order_amount": amount,
                     "delivery_minutes": deliv})
out = []
for k, r in enumerate(rows):
    if k % 100 == 7 and r["status"] == "completed" and r["delivery_minutes"] != "":
        r["delivery_minutes"] = -10                          # impossible negative duration (planted)
    if r["status"] == "completed" and rng.random() < .01:
        r["order_amount"] = "N/A" if rng.random() < .5 else "error"   # unparseable amount (planted)
    base = [r["order_id"], r["customer_id"], r["product_id"], r["quantity"], r["order_ts"].strftime("%Y-%m-%d %H:%M:%S"),
            r["status"], r["discount_pct"], r["order_amount"], r["delivery_minutes"]]
    out.append(base)
    if rng.random() < .05:   # earlier 'pending' record of the same order (partial duplicate)
        early = list(base); early[4] = (r["order_ts"] - timedelta(minutes=rng.randint(1, 10))).strftime("%Y-%m-%d %H:%M:%S")
        early[5] = "pending"; early[8] = ""
        out.append(early)
    if rng.random() < .01:   # exact duplicate
        out.append(list(base))
rng.shuffle(out)
write("orders.csv", ["order_id", "customer_id", "product_id", "quantity", "order_ts", "status",
                     "discount_pct", "order_amount", "delivery_minutes"], out)

# ---------------- support tickets ----------------
T = {"billing": ["I was charged twice for order {o}. Please refund the extra payment. My email is {e}.",
                 "The invoice for {o} shows a wrong amount. Call me on {p} to correct the bill.",
                 "Extra money was deducted for order {o}, please check the bill. Contact {e}."],
     "delivery": ["My order {o} has not arrived and the rider is not picking up calls. Please call {p}.",
                  "Delivery for {o} is very late, it was promised in 20 minutes. Write to {e}.",
                  "The rider marked {o} as delivered but I never received the package. Call {p}."],
     "technical": ["The app crashes every time I open the cart. I tried to place {o} again and got an error. Email {e}.",
                   "I cannot login, the OTP never arrives on {p}.",
                   "The payment page shows an error and the app freezes when I apply a coupon. Order {o}."],
     "fraud": ["There is an unauthorized transaction on my card linked to order {o}. I did not place it. Call {p}.",
               "Someone hacked my account and used my wallet balance. This is fraud, contact {e}.",
               "I was charged for order {o} that I never made, it looks like fraud on my account. Call {p}."]}
order_by_cust = {}
for r in out:
    order_by_cust.setdefault(r[1], r[0])
cust_with_orders = sorted(order_by_cust)
tickets = []
cats = list(T)
for n in range(1, 157):
    cat = cats[(n - 1) % 4]
    cid = rng.choice(cust_with_orders)
    txt = rng.choice(T[cat]).format(o=order_by_cust[cid], e=f"user{n}@example.com", p=str(rng.choice("6789")) + "".join(rng.choice("0123456789") for _ in range(9)))
    tickets.append([f"T{n:04d}", cid, order_by_cust[cid], txt, cat])
FILL = "The same problem keeps happening and nobody has given me a clear answer yet. "
for n in range(157, 161):                         # 4 empty tickets
    tickets.append([f"T{n:04d}", rng.choice(cust_with_orders), "", "", "none"])
for n in range(161, 165):                         # 4 very long tickets (~520 words)
    tickets.append([f"T{n:04d}", rng.choice(cust_with_orders), "", ("My delivery was late. " + FILL * 40).strip(), "delivery"])
rng.shuffle(tickets)
write("tickets.csv", ["ticket_id", "customer_id", "order_id", "text", "true_category"], tickets)

# ---------------- policy documents ----------------
DOCS = {
"refund_policy.txt": """Refund Policy. A refund can be requested only for orders whose status is completed. Orders with status cancelled were never charged, so no refund applies to them. If a completed order is delivered in more than 45 minutes, the customer receives a full refund of the order amount. If a completed order is delivered in 45 minutes or less, no late-delivery refund is paid. Refunds up to INR 1000 are approved automatically by the support system. A refund above INR 1000 must be approved by a human support lead before money is returned. Approved refunds are credited to the original payment method within 5 working days. A refund request must be raised within 7 days of the delivery date.""",
"delivery_policy.txt": """Delivery Policy. Standard delivery is promised within 30 minutes of order confirmation in all supported cities. The supported cities are Chennai, Mumbai, Bengaluru, Delhi, Hyderabad and Pune. A delivery fee of INR 25 is charged on orders below INR 199, and delivery is free on orders of INR 199 or more. During heavy rain the promised time may extend to 60 minutes and the app shows a rain notice. A rider will attempt to contact the customer twice before marking an order as undeliverable. Electronics items are delivered only after the customer shares a one time password with the rider.""",
"payment_policy.txt": """Payment Policy. Customers can pay using UPI, debit card, credit card or cash on delivery. Cash on delivery is available only for orders up to INR 2500. If a payment fails but money is deducted, the amount is returned automatically within 48 hours. A customer who sees two charges for one order should raise a billing ticket with the order identifier. Wallet balance can be used for any order and is never refunded as cash. Coupons cannot be combined with wallet balance on the same order.""",
"privacy_policy.txt": """Privacy Policy. Customer phone numbers and email addresses are used only for order updates and support. Support tickets are processed with personal details masked before any automated tool reads them. Delivery addresses are deleted from active systems 180 days after the last order. A customer can ask for account deletion and the request is completed within 30 days. Payment card numbers are never stored by the support system.""",
"escalation_policy.txt": """Escalation Policy. A ticket about fraud or an unauthorized transaction is marked high priority and sent to the security team within 2 hours. Tickets from premium customers are answered within 1 hour, and all other tickets within 24 hours. If an automated assistant cannot resolve a ticket within 6 steps, it must hand the case to a human agent with a short summary. A human agent must call the customer back if such a high-value request is rejected."""}
for fn, body in DOCS.items():
    with open(os.path.join(POL, fn), "w", encoding="utf-8") as f:
        f.write(body + "\n")

# ---------------- policy evaluation questions ----------------
QA = [("What delivery time makes a completed order eligible for a full refund?", "refund_policy.txt", "more than 45 minutes"),
      ("Who must approve a refund above INR 1000?", "refund_policy.txt", "human support lead"),
      ("How many days do I have to raise a refund request?", "refund_policy.txt", "within 7 days"),
      ("What is the delivery fee on small orders?", "delivery_policy.txt", "INR 25"),
      ("On orders of what value is delivery free?", "delivery_policy.txt", "INR 199 or more"),
      ("Why does the electronics delivery need a one time password?", "delivery_policy.txt", "one time password"),
      ("Is cash on delivery available for a large order?", "payment_policy.txt", "INR 2500"),
      ("How soon is money returned when a payment fails but is deducted?", "payment_policy.txt", "48 hours"),
      ("Can wallet balance be refunded as cash?", "payment_policy.txt", "never refunded as cash"),
      ("When are delivery addresses deleted from active systems?", "privacy_policy.txt", "180 days"),
      ("How fast must a fraud ticket reach the security team?", "escalation_policy.txt", "within 2 hours"),
      ("After how many steps must an automated assistant hand over to a human?", "escalation_policy.txt", "6 steps"),
      ("What is the favourite colour of the company mascot?", "NONE", ""),
      ("Who won the cricket world cup final?", "NONE", "")]
with open(os.path.join(RAW, "policy_eval.csv"), "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f); w.writerow(["question", "expected_doc", "expected_fact"]); w.writerows(QA)

# ---------------- agent scenarios (simulated refund requests) ----------------
from collections import Counter
cnt = Counter(r[0] for r in out)
def clean_row(r):                                    # single-record, fully valid order
    try:
        return cnt[r[0]] == 1 and r[5] in ("completed", "cancelled") and float(r[7]) > 0 and \
            (r[5] == "cancelled" or (r[8] != "" and 8 <= int(r[8]) <= 75))
    except ValueError:
        return False
ok_rows = sorted((r for r in out if clean_row(r)), key=lambda r: r[0])
def first(pred):
    return next(r for r in ok_rows if pred(r))
late_small = [r for r in ok_rows if r[5] == "completed" and int(r[8]) > 45 and float(r[7]) <= 1000][:2]
late_big = first(lambda r: r[5] == "completed" and int(r[8]) > 45 and float(r[7]) > 1000)
on_time = first(lambda r: r[5] == "completed" and int(r[8]) <= 45)
cancelled = first(lambda r: r[5] == "cancelled")
SC = []
for r in late_small:
    SC.append((r[1], f"My order {r[0]} reached me very late. I would like a refund.", "REFUND", float(r[7])))
SC.append((late_big[1], f"Order {late_big[0]} arrived far too late, please refund me.", "ESCALATE", float(late_big[7])))
SC.append((on_time[1], f"Please refund order {on_time[0]}, I did not like the delivery.", "DENIED", 0))
SC.append((cancelled[1], f"I want my money back for order {cancelled[0]}.", "DENIED", 0))
SC.append(("C0001", "Refund for order ORD-999999 please.", "NOT_FOUND", 0))
SC.append((late_small[0][1], f"[LOOP] Refund order {late_small[0][0]}.", "FALLBACK", 0))
with open(os.path.join(RAW, "agent_scenarios.csv"), "w", newline="", encoding="utf-8") as f:
    w = csv.writer(f); w.writerow(["scenario_id", "customer_id", "message", "expected_outcome", "expected_refund_inr"])
    for i, (cid, msg, oc, amt) in enumerate(SC, 1):
        w.writerow([f"S{i}", cid, msg, oc, f"{amt:.2f}"])
print("generated: customers=%d products=%d orders_raw=%d tickets=%d" % (len(cust), len(PRODUCTS), len(out), len(tickets)))
```

```python
import json, math, re
from types import SimpleNamespace as NS

class TransientLLMError(Exception):
    """Simulated temporary failure (timeout / server busy). Retrying the same call succeeds."""

_state = {"calls": 0, "ok": 0}      # call counters that drive the simulated misbehaviour

def reset():
    """Restart the counters. Call once before each independent batch run."""
    _state["calls"] = 0; _state["ok"] = 0

CATS = ["billing", "delivery", "technical", "fraud"]
KW = {"fraud": {"unauthorized", "fraud", "hacked", "stolen"},
      "billing": {"charged", "invoice", "bill", "refund", "deducted", "payment"},
      "technical": {"app", "crash", "login", "otp", "error", "freezes"},
      "delivery": {"arrived", "rider", "delivery", "delivered", "late"}}

def est_tokens(text):
    return math.ceil(len((text or "").split()) * 1.3)

def _words(t):
    return set(re.findall(r"[a-z]+", (t or "").lower()))

def _resp(content=None, tool_calls=None, finish="stop", messages=()):
    prompt = sum(est_tokens(m.get("content") or json.dumps(m.get("tool_calls", ""), default=str)) for m in messages)
    comp = est_tokens(content or json.dumps(tool_calls or "", default=str))
    msg = NS(role="assistant", content=content, tool_calls=[NS(**tc) for tc in (tool_calls or [])])
    return NS(choices=[NS(message=msg, finish_reason=finish)],
              usage=NS(prompt_tokens=prompt, completion_tokens=comp, total_tokens=prompt + comp))

def _classify(messages):
    system, user = messages[0]["content"], messages[-1]["content"]
    _state["calls"] += 1
    if _state["calls"] % 9 == 0:                                 # every 9th call fails once
        raise TransientLLMError("503 server busy")
    _state["ok"] += 1
    order = ["fraud", "technical", "billing", "delivery"] if "EXAMPLES:" in system else ["billing", "fraud", "technical", "delivery"]
    w = _words(user)
    cat = next((c for c in order if KW[c] & w), "technical")
    if _state["ok"] % 13 == 0:                                   # chatty, non-JSON answer
        return _resp(f"The customer appears to have a {cat}-related problem.", messages=messages)
    if _state["ok"] % 17 == 0:                                   # JSON, but category outside the allowed list
        return _resp(json.dumps({"category": "shipping", "reason": "parcel movement"}), messages=messages)
    return _resp(json.dumps({"category": cat, "reason": "keyword match"}), messages=messages)

STOP = {"the", "a", "an", "is", "of", "to", "for", "in", "on", "how", "what", "when", "who", "are", "do", "does",
        "i", "my", "be", "can", "after", "from", "it", "or", "and", "at", "by", "which", "why", "if", "than", "that"}

def _stems(t):
    return {w[:5] for w in re.findall(r"[a-z]+", (t or "").lower()) if w not in STOP}

def _policy(messages):
    """Answers with the context sentence that best matches the question (rare words weigh more)."""
    user = messages[-1]["content"]
    m = re.search(r"CONTEXT:\s*(.*?)\s*QUESTION:\s*(.*)", user, re.S)
    context, question = (m.group(1), m.group(2)) if m else ("", user)
    sents = [s.strip() for s in re.split(r"(?<=[.])\s+", re.sub(r"\[[^\]]*\.txt\]", " ", context)) if s.strip()]
    qs = _stems(question)
    df = {q: sum(q in _stems(s) for s in sents) for q in qs}
    best, best_score = None, 0.0
    for s in sents:
        sc = sum(1.0 / df[q] for q in qs & _stems(s))
        if sc > best_score:
            best, best_score = s, sc
    if best is None:
        return _resp("I could not find that information in the policy documents.", messages=messages)
    return _resp(best, messages=messages)

def _agent(messages, tools):
    user = next(m["content"] for m in messages if m["role"] == "user")
    tool_msgs = [m for m in messages if m["role"] == "tool"]
    call = lambda n, a: _resp(None, [{"id": f"call_{len(tool_msgs)+1}", "name": n, "arguments": a}], "tool_calls", messages)
    if "[LOOP]" in user:
        return call("search_policy", {"query": "refund policy"})
    if not tool_msgs:
        m = re.search(r"ORD-\d{6}", user)
        return call("get_order", {"order_id": m.group(0)}) if m else _resp("NEEDS_INFO: please share the order id.", messages=messages)
    last = tool_msgs[-1]; data = json.loads(last["content"]); name = last["name"]
    if name == "get_order":
        if "error" in data:
            return _resp("NOT_FOUND: no such order.", messages=messages)
        return call("search_policy", {"query": "refund policy late delivery"})
    order = json.loads(next(m for m in tool_msgs if m["name"] == "get_order")["content"])
    if name == "search_policy":
        if order["status"] != "completed":
            return _resp("DENIED: only completed orders can be refunded.", messages=messages)
        return call("calculate_refund", {"order_amount": order["order_amount"], "delivery_minutes": order["delivery_minutes"]})
    if name == "calculate_refund":
        amt = data["refund_inr"]
        if amt == 0:
            return _resp("DENIED: delivery was within 45 minutes.", messages=messages)
        if amt > 1000:
            return call("escalate_to_human", {"order_id": order["order_id"], "refund_inr": amt})
        return call("issue_refund", {"order_id": order["order_id"], "refund_inr": amt})
    if name == "issue_refund":
        return _resp(f"REFUND: INR {data['refund_inr']} issued.", messages=messages)
    if name == "escalate_to_human":
        return _resp("ESCALATE: refund above INR 1000 sent to a human lead.", messages=messages)
    return _resp("DENIED: unable to process.", messages=messages)

def chat(messages, tools=None):
    head = messages[0]["content"].upper()
    if "CLASSIFY" in head:
        return _classify(messages)
    if "POLICY ASSISTANT" in head:
        return _policy(messages)
    if "AGENT" in head:
        return _agent(messages, tools)
    raise ValueError("first message must name one of: CLASSIFY, POLICY ASSISTANT, AGENT")
```

```python
import os
import sqlite3
import numpy as np
import pandas as pd

def run_pipeline():
    os.makedirs("data/clean", exist_ok=True)
    os.makedirs("outputs", exist_ok=True)
    os.makedirs("sql", exist_ok=True)

    # 1. Load raw data
    raw_cust = pd.read_csv("data/raw/customers.csv")
    raw_prod = pd.read_csv("data/raw/products.csv")
    raw_orders = pd.read_csv("data/raw/orders.csv")

    # --- Clean Customers ---
    cust = raw_cust.copy()
    # Standardize city: strip whitespace and apply title case
    cust["city"] = cust["city"].str.strip().str.title()

    # Map gender variations to male, female, or unknown
    def clean_gender(g):
        if pd.isna(g) or str(g).strip() == "":
            return "unknown"
        g_lower = str(g).strip().lower()
        if g_lower in ["m", "male"]:
            return "male"
        elif g_lower in ["f", "female"]:
            return "female"
        return "unknown"
    cust["gender"] = cust["gender"].apply(clean_gender)

    # Treat ages outside 10-100 as missing, then fill with city median
    cust["age"] = pd.to_numeric(cust["age"], errors="coerce")
    cust.loc[(cust["age"] < 10) | (cust["age"] > 100), "age"] = np.nan
    cust["age"] = cust["age"].fillna(cust.groupby("city")["age"].transform("median"))
    cust["age"] = cust["age"].round().astype(int)

    cust["signup_date"] = pd.to_datetime(cust["signup_date"])

    # --- Clean Orders ---
    orders = raw_orders.copy()
    orders["order_ts"] = pd.to_datetime(orders["order_ts"])

    # Remove exact duplicate rows
    orders = orders.drop_duplicates()

    # For each order_id, keep only the record with the latest order_ts
    orders = orders.sort_values("order_ts").drop_duplicates(subset=["order_id"], keep="last")

    # Convert order_amount to numeric; fill unparseable entries using product price, quantity, and discount
    orders["order_amount"] = pd.to_numeric(orders["order_amount"], errors="coerce")

    # Merge with products to get price
    orders = orders.merge(raw_prod[["product_id", "price_inr"]], on="product_id", how="left")
    missing_amt = orders["order_amount"].isna()
    calc_amount = (orders["price_inr"] * orders["quantity"] * (1 - orders["discount_pct"] / 100)).round(2)
    orders.loc[missing_amt, "order_amount"] = calc_amount[missing_amt]
    orders = orders.drop(columns=["price_inr"])

    # Handle delivery_minutes
    orders["delivery_minutes"] = pd.to_numeric(orders["delivery_minutes"], errors="coerce")
    orders.loc[orders["delivery_minutes"] < 0, "delivery_minutes"] = np.nan

    # Impute missing delivery_minutes for completed orders using category median among completed orders
    prod_cat = raw_prod.set_index("product_id")["category"].to_dict()
    orders["category"] = orders["product_id"].map(prod_cat)

    completed_mask = (orders["status"] == "completed")
    cat_medians = orders[completed_mask].groupby("category")["delivery_minutes"].median()

    for cat, med in cat_medians.items():
        fill_mask = completed_mask & (orders["category"] == cat) & (orders["delivery_minutes"].isna())
        orders.loc[fill_mask, "delivery_minutes"] = med

    orders = orders.drop(columns=["category"])

    # Save clean CSVs
    cust.to_csv("data/clean/customers.csv", index=False)
    raw_prod.to_csv("data/clean/products.csv", index=False)
    orders.to_csv("data/clean/orders.csv", index=False)

    # Store in SQLite database
    db_path = "data/instamart.db"
    if os.path.exists(db_path):
        os.remove(db_path)
    conn = sqlite3.connect(db_path)
    cust.to_sql("customers", conn, index=False, if_exists="replace")
    raw_prod.to_sql("products", conn, index=False, if_exists="replace")
    # Format order_ts as text YYYY-MM-DD HH:MM:SS
    orders_db = orders.copy()
    orders_db["order_ts"] = orders_db["order_ts"].dt.strftime("%Y-%m-%d %H:%M:%S")
    orders_db.to_sql("orders", conn, index=False, if_exists="replace")
    conn.close()
    print("Data cleaning and database storage completed successfully.")

if __name__ == "__main__":
    run_pipeline()
```

```python
SELECT customer_id, SUM(order_amount) as total_spend
FROM orders
WHERE status = 'completed' AND order_amount > 500
GROUP BY customer_id
HAVING SUM(order_amount) > 8000;

-- Q2: Three-table join (Revenue by city and product category)
SELECT c.city, p.category, SUM(o.order_amount) AS revenue
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
JOIN products p ON o.product_id = p.product_id
WHERE o.status = 'completed'
GROUP BY c.city, p.category
ORDER BY revenue DESC;

-- Q3: Scalar subquery (Products priced above average catalogue price)
SELECT product_id, product_name, price_inr
FROM products
WHERE price_inr > (SELECT AVG(price_inr) FROM products);

-- Q4: LEFT JOIN (Customers who have no completed order)
SELECT c.customer_id
FROM customers c
LEFT JOIN orders o ON c.customer_id = o.customer_id AND o.status = 'completed'
WHERE o.order_id IS NULL;

-- Q5: Common Table Expression (Customers whose total completed spend is above average spend)
WITH customer_spend AS (
    SELECT customer_id, SUM(order_amount) AS total_spend
    FROM orders
    WHERE status = 'completed'
    GROUP BY customer_id
),
avg_spend AS (
    SELECT AVG(total_spend) AS avg_total FROM customer_spend
)
cs.customer_id, cs.total_spend
FROM customer_spend cs, avg_spend as
WHERE cs.total_spend > as.avg_total;

-- Q6: Ranking window functions (Rank, Dense_Rank, Row_Number partitioned by city)
WITH cust_city_spend AS (
    SELECT c.customer_id, c.city, SUM(o.order_amount) AS total_spend
    FROM orders o
    JOIN customers c ON o.customer_id = c.customer_id
    WHERE o.status = 'completed'
    GROUP BY c.customer_id, c.city
)
SELECT customer_id, city, total_spend,
       RANK() OVER (PARTITION BY city ORDER BY total_spend DESC) as rnk,
       DENSE_RANK() OVER (PARTITION BY city ORDER BY total_spend DESC) as dense_rnk,
       ROW_NUMBER() OVER (PARTITION BY city ORDER BY total_spend DESC) as row_num
FROM cust_city_spend;

-- Q7: Running total (Daily completed cumulative revenue)
WITH daily_revenue AS (
    SELECT SUBSTR(order_ts, 1, 10) AS day, SUM(order_amount) AS daily_rev
    FROM orders
    WHERE status = 'completed'
    GROUP BY day
)
SELECT day, daily_rev,
       SUM(daily_rev) OVER (ORDER BY day ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW) AS cumulative_revenue
FROM daily_revenue;
```

```python
import os
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import f1_score, precision_score, recall_score, confusion_matrix
import joblib

def run_part_b():
    os.makedirs("outputs", exist_ok=True)

    # 1. Load clean data
    orders = pd.read_csv("data/clean/orders.csv")
    customers = pd.read_csv("data/clean/customers.csv")

    orders["order_ts"] = pd.to_datetime(orders["order_ts"])
    snapshot = pd.to_datetime("2025-10-01 00:00:00")

    # Filter completed orders for RFM
    comp_orders = orders[orders["status"] == "completed"]

    rfm = comp_orders.groupby("customer_id").agg(
        recency=("order_ts", lambda x: (snapshot - x.max()).days),
        frequency=("order_id", "count"),
        monetary=("order_amount", "sum")
    ).reset_index()

    # Ensure all customers with orders are included
    all_cust_with_orders = customers[customers["customer_id"].isin(comp_orders["customer_id"])].copy()
    rfm_full = all_cust_with_orders[["customer_id"]].merge(rfm, on="customer_id", how="left").fillna({"recency": 183, "frequency": 0, "monetary": 0})

    # 2. K-means Clustering (K=4 chosen based on silhouette & elbow analysis)
    scaler = StandardScaler()
    rfm_scaled = scaler.fit_transform(rfm_full[["recency", "frequency", "monetary"]])

    kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)
    rfm_full["cluster"] = kmeans.fit_predict(rfm_scaled)

    # Profile clusters to name segments
    profile = rfm_full.groupby("cluster")[["recency", "frequency", "monetary"]].mean()
    print("Cluster Profiles:\n", profile)

    # Identify premium segment (highest mean monetary)
    prem_cluster = profile["monetary"].idxmax()

    def assign_segment(c):
        if c == prem_cluster:
            return "Premium", True
        elif c == 1:
            return "Frequent", False
        elif c == 2:
            return "At-Risk", False
        else:
            return "Occasional", False

    rfm_full[["segment_name", "is_premium"]] = rfm_full["cluster"].apply(lambda x: pd.Series(assign_segment(x)))
    rfm_full["segment_id"] = rfm_full["cluster"]

    output_seg = rfm_full[["customer_id", "recency", "frequency", "monetary", "segment_id", "segment_name", "is_premium"]]
    output_seg.to_csv("outputs/customer_segments.csv", index=False)

    # PCA
    pca = PCA(n_components=2, random_state=42)
    pca.fit(rfm_scaled)
    print("PCA explained variance ratio:", pca.explained_variance_ratio_)

    # 3. Lapse Prediction (Cutoff: 2025-09-01)
    cutoff = pd.to_datetime("2025-09-01 00:00:00")
    pre_orders = orders[orders["order_ts"] < cutoff]

    cohort_cids = pre_orders[pre_orders["status"] == "completed"]["customer_id"].unique()
    cohort_cust = customers[customers["customer_id"].isin(cohort_cids)].copy()

    # Features from pre-cutoff orders
    pre_comp = pre_orders[pre_orders["status"] == "completed"]
    cust_feat = pre_comp.groupby("customer_id").agg(
        recency=("order_ts", lambda x: (cutoff - x.max()).days),
        frequency=("order_id", "count"),
        monetary=("order_amount", "sum"),
        avg_discount=("discount_pct", "mean"),
        avg_delivery=("delivery_minutes", "mean")
    ).reset_index()

    # Cancellation rate before cutoff
    all_pre_status = pre_orders.groupby("customer_id")["status"].agg(
        total=lambda x: (x == "completed").sum() + (x == "cancelled").sum(),
        cancelled=lambda x: (x == "cancelled").sum()
    ).reset_index()
    all_pre_status["cancel_rate"] = all_pre_status["cancelled"] / all_pre_status["total"].replace(0, 1)

    # Favorite product category
    prod_cat = pd.read_csv("data/clean/products.csv")
    pre_orders_prod = pre_orders.merge(prod_cat, on="product_id", how="left")
    fav_cat = pre_orders_prod.groupby("customer_id")["category"].agg(lambda x: x.mode()[0] if not x.empty else "Grocery").reset_index(name="fav_category")

    # Tenure
    cohort_cust["tenure"] = (cutoff - cohort_cust["signup_date"]).dt.days

    model_df = cohort_cust.merge(cust_feat, on="customer_id", how="left") \
                          .merge(all_pre_status[["customer_id", "cancel_rate"]], on="customer_id", how="left") \
                          .merge(fav_cat, on="customer_id", how="left")

    model_df = model_df.fillna({"recency": (cutoff - pd.to_datetime("2025-04-01")).days, "frequency": 0, "monetary": 0, "cancel_rate": 0, "avg_delivery": 24, "avg_discount": 0})

    # Define Label: lapsed = 1 if no completed order between 2025-09-01 and 2025-09-30
    sep_orders = orders[(orders["order_ts"] >= cutoff) & (orders["order_ts"] <= pd.to_datetime("2025-09-30 23:59:59")) & (orders["status"] == "completed")]
    active_in_sep = set(sep_orders["customer_id"].unique())
    model_df["lapsed"] = model_df["customer_id"].apply(lambda cid: 0 if cid in active_in_sep else 1)

    features = ["recency", "frequency", "monetary", "avg_discount", "cancel_rate", "avg_delivery", "fav_category", "city", "age", "gender", "tenure"]
    X = model_df[features]
    y = model_df["lapsed"]

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)

    # Preprocessing pipeline setup
    from sklearn.compose import ColumnTransformer
    from sklearn.pipeline import Pipeline
    from sklearn.impute import SimpleImputer
    from sklearn.preprocessing import OneHotEncoder

    numeric_features = ["recency", "frequency", "monetary", "avg_discount", "cancel_rate", "avg_delivery", "age", "tenure"]
    categorical_features = ["fav_category", "city", "gender"]

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', Pipeline([('imputer', SimpleImputer(strategy='median')), ('scaler', StandardScaler())]), numeric_features),
            ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_features)
        ]
    )

    # Logistic Regression tuning
    lr_pipeline = Pipeline([('prep', preprocessor), ('clf', LogisticRegression(max_iter=1000, random_state=42))])
    lr_grid = GridSearchCV(lr_pipeline, {"clf__C": [0.01, 0.1, 1, 10, 100]}, cv=5, scoring="f1")
    lr_grid.fit(X_train, y_train)
    print("Logistic Regression Best Params:", lr_grid.best_params_, "F1:", lr_grid.best_score_)

    # Random Forest tuning
    rf_pipeline = Pipeline([('prep', preprocessor), ('clf', RandomForestClassifier(n_estimators=100, random_state=42))])
    rf_grid = GridSearchCV(rf_pipeline, {
        "clf__max_depth": [3, 5, 8],
        "clf__min_samples_leaf": [5, 10, 20]
    }, cv=5, scoring="f1")
    rf_grid.fit(X_train, y_train)
    print("Random Forest Best Params:", rf_grid.best_params_, "F1:", rf_grid.best_score_)

    best_model = rf_grid.best_estimator_
    joblib.dump(best_model, "outputs/lapse_model.joblib")

    # Retention list for test Premium customers
    test_idx = X_test.index
    test_df = model_df.loc[test_idx].copy()
    test_df["lapse_probability"] = best_model.predict_proba(X_test)[:, 1]

    prem_test = test_df.merge(output_seg[["customer_id", "is_premium"]], on="customer_id", how="left")
    retention_list = prem_test[prem_test["is_premium"] == True].sort_values("lapse_probability", ascending=False).head(10)
    retention_list[["customer_id", "lapse_probability", "lapsed"]].rename(columns={"lapsed": "actually_lapsed"}).to_csv("outputs/retention_list.csv", index=False)
    print("Part B completed successfully.")

if __name__ == "__main__":
    run_part_b()
```

```python
import os
import re
import json
import random
import time
import pandas as pd
from mock_llm import chat, reset, est_tokens, TransientLLMError

MAX_INPUT_TOKENS = 400
BASE_DELAY = 0.05

EMAIL_REGEX = r'[\w.+-]+@[\w-]+\.[\w.]+'
PHONE_REGEX = r'\b[6-9]\d{9}\b'
ORDER_REGEX = r'ORD-\d{6}'

ALLOWED_CATEGORIES = {"billing", "delivery", "technical", "fraud"}

ZERO_SHOT_PROMPT = (
    "CLASSIFY: You are a support ticket classifier. "
    "Allowed categories: billing, delivery, technical, fraud. "
    "Reply ONLY with valid JSON in this exact form: {\"category\": \"...\", \"reason\": \"...\"}."
)

FEW_SHOT_PROMPT = (
    "CLASSIFY: You are a support ticket classifier. "
    "Allowed categories: billing, delivery, technical, fraud. "
    "Reply ONLY with valid JSON in this exact form: {\"category\": \"...\", \"reason\": \"...\"}\n"
    "EXAMPLES:\n"
    "Ticket: Unauthorized transaction on my account for order ORD-123456.\n"
    "Response: {\"category\": \"fraud\", \"reason\": \"unauthorized transaction\"}\n"
    "Ticket: My app crashes every time I open the cart.\n"
    "Response: {\"category\": \"technical\", \"reason\": \"app crash\"}"
)

def redact_text(text):
    if not isinstance(text, str):
        return "", 0, 0, 0
    t_text = text
    t_text, e_count = re.subn(EMAIL_REGEX, "[EMAIL]", t_text)
    t_text, p_count = re.subn(PHONE_REGEX, "[PHONE]", t_text)
    t_text, o_count = re.subn(ORDER_REGEX, "[ORDER_ID]", t_text)
    total_sub = e_count + p_count + o_count
    return t_text, e_count, p_count, o_count

def resilient_call(messages):
    for n in range(3):
        try:
            return chat(messages)
        except TransientLLMError as e:
            wait = BASE_DELAY * (2 ** n) * random.uniform(0.9, 1.15)
            print(f"Retry {n+1}/3 due to TransientLLMError: {e}. Waiting {wait:.4f}s")
            time.sleep(wait)
    raise RuntimeError("Failed after 3 retries due to persistent TransientLLMError")

def validate_response(resp_content, raw_text):
    # Try parsing JSON
    try:
        data = json.loads(resp_content)
        cat = data.get("category")
        if cat in ALLOWED_CATEGORIES:
            return "classified_json", cat
    except json.JSONDecodeError:
        pass

    # Fallback keyword extraction using re.findall
    words = re.findall(r"[a-z]+", (resp_content or "").lower())
    matches = [c for c in ALLOWED_CATEGORIES if c in words]
    if len(matches) == 1:
        return "classified_keyword", matches[0]

    return "flagged_invalid_output", None

def run_triage():
    os.makedirs("outputs", exist_ok=True)
    tickets_df = pd.read_csv("data/raw/tickets.csv", keep_default_na=False)
    segments_df = pd.read_csv("outputs/customer_segments.csv")
    premium_set = set(segments_df[segments_df["is_premium"] == True]["customer_id"])

    # 1. PII Redaction Audit across all tickets
    total_e, total_p, total_o = 0, 0, 0
    redacted_texts = []
    for _, row in tickets_df.iterrows():
        rt, e, p, o = redact_text(row["text"])
        total_e += e
        total_p += p
        total_o += o
        # Assert no PII remains
        assert not re.search(EMAIL_REGEX, rt), "Email leak found"
        assert not re.search(PHONE_REGEX, rt), "Phone leak found"
        assert not re.search(ORDER_REGEX, rt), "Order ID leak found in redacted text"
        redacted_texts.append(rt)

    print(f"PII Redaction Totals -> Emails: {total_e}, Phones: {total_p}, Order IDs: {total_o}, Total: {total_e+total_p+total_o}")
    tickets_df["redacted_text"] = redacted_texts

    # 2. Run Few-Shot Pipeline for final outputs/triage_results.csv
    reset()
    results = []
    running_tokens = 0
    BUDGET = 100000

    for _, row in tickets_df.iterrows():
        tid = row["ticket_id"]
        cid = row["customer_id"]
        text = row["text"]
        redacted = row["redacted_text"]

        # Input guardrails
        if not text or text.strip() == "":
            results.append({
                "ticket_id": tid, "customer_id": cid, "status": "skipped_empty",
                "category": "", "segment_name": "", "priority": "NORMAL", "route": "human_review", "tokens_used": 0
            })
            continue

        if est_tokens(text) > MAX_INPUT_TOKENS:
            results.append({
                "ticket_id": tid, "customer_id": cid, "status": "flagged_long",
                "category": "", "segment_name": "", "priority": "NORMAL", "route": "human_review", "tokens_used": 0
            })
            continue

        if running_tokens >= BUDGET:
            results.append({
                "ticket_id": tid, "customer_id": cid, "status": "budget_exhausted",
                "category": "", "segment_name": "", "priority": "NORMAL", "route": "human_review", "tokens_used": 0
            })
            continue

        messages = [
            {"role": "system", "content": FEW_SHOT_PROMPT},
            {"role": "user", "content": redacted}
        ]

        try:
            resp = resilient_call(messages)
            toks = resp.usage.total_tokens
            running_tokens += toks
            status, cat = validate_response(resp.choices[0].message.content, resp.choices[0].message.content)
        except Exception:
            status, cat = "flagged_invalid_output", None
            toks = 0

        # Segment lookup for routing
        seg_row = segments_df[segments_df["customer_id"] == cid]
        seg_name = seg_row["segment_name"].values[0] if not seg_row.empty else ""
        is_prem = cid in premium_set

        # Routing rules
        if status in ["skipped_empty", "flagged_long", "flagged_invalid_output"]:
            route, priority = "human_review", "NORMAL"
        elif cat == "fraud":
            route, priority = "security_team", "HIGH"
        elif is_prem:
            route, priority = "premium_desk", "HIGH"
        else:
            route, priority = "support_queue", "NORMAL"

        results.append({
            "ticket_id": tid, "customer_id": cid, "status": status,
            "category": cat if cat else "", "segment_name": seg_name,
            "priority": priority, "route": route, "tokens_used": toks
        })

    triage_res_df = pd.DataFrame(results)
    triage_res_df.to_csv("outputs/triage_results.csv", index=False)

    # 3. Budget Demo run (Zero-shot with TOKEN_BUDGET = 6000)
    reset()
    budget_results = []
    b_tokens = 0
    DEMO_BUDGET = 6000

    for _, row in tickets_df.iterrows():
        tid = row["ticket_id"]
        cid = row["customer_id"]
        text = row["text"]
        redacted = row["redacted_text"]

        if not text or text.strip() == "":
            budget_results.append({"ticket_id": tid, "status": "skipped_empty"})
            continue
        if est_tokens(text) > MAX_INPUT_TOKENS:
            budget_results.append({"ticket_id": tid, "status": "flagged_long"})
            continue
        if b_tokens >= DEMO_BUDGET:
            budget_results.append({"ticket_id": tid, "status": "budget_exhausted"})
            continue

        messages = [
            {"role": "system", "content": ZERO_SHOT_PROMPT},
            {"role": "user", "content": redacted}
        ]
        try:
            resp = resilient_call(messages)
            b_tokens += resp.usage.total_tokens
            status, _ = validate_response(resp.choices[0].message.content, resp.choices[0].message.content)
        except Exception:
            status = "flagged_invalid_output"
        budget_results.append({"ticket_id": tid, "status": status})

    pd.DataFrame(budget_results).to_csv("outputs/triage_budget_demo.csv", index=False)
    print("Part C pipeline executed successfully.")

if __name__ == "__main__":
    run_triage()
```

```python
import re
import json
import time
import math
from datetime import datetime
import sqlite3
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from mock_llm import chat, reset, est_tokens, TransientLLMError

# 1. Chunking and TF-IDF Retrieval
POL_DIR = "data/policies"
chunks = []
for fn in sorted(os.listdir(POL_DIR)):
    if fn.endswith(".txt"):
        path = os.path.join(POL_DIR, fn)
        with open(path, "r", encoding="utf-8") as f:
            text = f.read().strip()
        words = text.split()
        step = 30
        chunk_size = 40
        for i in range(0, len(words), step):
            chunk_words = words[i:i+chunk_size]
            if not chunk_words:
                break
            chunks.append({"source": fn, "text": " ".join(chunk_words)})
            if i + chunk_size >= len(words):
                break

chunk_texts = [c["text"] for c in chunks]
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(chunk_texts)

def retrieve(query, k=3):
    q_vec = vectorizer.transform([query])
    scores = cosine_similarity(q_vec, tfidf_matrix).flatten()
    top_indices = scores.argsort()[::-1][:k]
    results = []
    for idx in top_indices:
        results.append({
            "source": chunks[idx]["source"],
            "score": float(scores[idx]),
            "text": chunks[idx]["text"]
        })
    return results

def answer_question(question):
    retrieved = retrieve(question, k=3)
    context_str = "\n".join([f"[{r['source']}] {r['text']}" for r in retrieved])
    user_msg = f"CONTEXT:\n{context_str}\nQUESTION:\n{question}"
    messages = [
        {"role": "system", "content": "POLICY ASSISTANT: Answer only from the supplied context and say you could not find the information otherwise."},
        {"role": "user", "content": user_msg}
    ]
    resp = chat(messages)
    return resp.choices[0].message.content, [r["source"] for r in retrieved]

# 2. Define Tools
def get_order(order_id):
    conn = sqlite3.connect("data/instamart.db")
    df = pd.read_sql_query("SELECT order_id, status, order_amount, delivery_minutes FROM orders WHERE order_id = ?", conn, params=(order_id,))
    conn.close()
    if df.empty:
        return {"error": "order not found"}
    row = df.iloc[0]
    return {
        "order_id": row["order_id"],
        "status": row["status"],
        "order_amount": float(row["order_amount"]),
        "delivery_minutes": None if pd.isna(row["delivery_minutes"]) else int(row["delivery_minutes"])
    }

def search_policy(query):
    return {"results": retrieve(query, k=3)}

def calculate_refund(order_amount, delivery_minutes):
    if delivery_minutes is None or delivery_minutes > 45:
        amt = float(order_amount)
    else:
        amt = 0.0
    return {"refund_inr": round(amt, 2)}

def log_ledger(filename, entry):
    os.makedirs("outputs", exist_ok=True)
    with open(os.path.join("outputs", filename), "a", encoding="utf-8") as f:
        f.write(json.dumps(entry) + "\n")

def issue_refund(order_id, refund_inr):
    entry = {"timestamp": datetime.now().isoformat(), "tool": "issue_refund", "order_id": order_id, "refund_inr": refund_inr}
    log_ledger("refund_ledger.jsonl", entry)
    return {"refund_inr": float(refund_inr)}

def escalate_to_human(order_id, refund_inr):
    entry = {"timestamp": datetime.now().isoformat(), "tool": "escalate_to_human", "order_id": order_id, "refund_inr": refund_inr}
    log_ledger("refund_ledger.jsonl", entry)
    return {"refund_inr": float(refund_inr)}

TOOL_MAP = {
    "get_order": get_order,
    "search_policy": search_policy,
    "calculate_refund": calculate_refund,
    "issue_refund": issue_refund,
    "escalate_to_human": escalate_to_human
}

# 3. ReAct Agent Loop with Limits and State Checkpoints
def run_agent(user_message, scenario_id, max_iterations=6, token_limit=6000, time_limit_seconds=10, run_id=None, crash_after_tools=0):
    start_time = time.monotonic()
    if not run_id:
        run_id = f"run_{scenario_id}_{int(time.time())}"

    messages = [
        {"role": "system", "content": "AGENT: You are a helpful refund agent that can use tools."},
        {"role": "user", "content": user_message}
    ]

    tokens_used = sum(est_tokens(m.get("content") or json.dumps(m.get("tool_calls", ""), default=str)) for m in messages)
    steps = []
    executed_tools = []

    for step_idx in range(max_iterations):
        if time.monotonic() - start_time >= time_limit_seconds:
            return {
                "scenario_id": scenario_id,
                "status": "FALLBACK_TIMEOUT",
                "final_answer": "FALLBACK_TIMEOUT: time limit exceeded.",
                "llm_calls": step_idx,
                "tokens_used": tokens_used,
                "steps": steps,
                "executed_tools": executed_tools
            }

        try:
            resp = chat(messages, tools=list(TOOL_MAP.keys()))
        except Exception:
            break

        tokens_used += resp.usage.total_tokens
        if tokens_used > token_limit:
            return {
                "scenario_id": scenario_id,
                "status": "FALLBACK_TOKEN_LIMIT",
                "final_answer": "FALLBACK_TOKEN_LIMIT: token limit exceeded.",
                "llm_calls": step_idx + 1,
                "tokens_used": tokens_used,
                "steps": steps,
                "executed_tools": executed_tools
            }

        msg = resp.choices[0].message
        if not msg.tool_calls:
            final_content = msg.content or ""
            status = final_content.split(":")[0] if ":" in final_content else "DENIED"
            return {
                "scenario_id": scenario_id,
                "status": status,
                "final_answer": final_content,
                "llm_calls": step_idx + 1,
                "tokens_used": tokens_used,
                "steps": steps,
                "executed_tools": executed_tools
            }

        tc_dicts = [{"id": tc.id, "name": tc.name, "arguments": tc.arguments} for tc in msg.tool_calls]
        messages.append({"role": "assistant", "content": msg.content, "tool_calls": tc_dicts})

        for tc in msg.tool_calls:
            t_name = tc.name
            t_args = json.loads(tc.arguments) if isinstance(tc.arguments, str) else tc.arguments

            if t_name in TOOL_MAP:
                obs = TOOL_MAP[t_name](**t_args)
            else:
                obs = {"error": "unknown tool"}

            executed_tools.append(t_name)
            obs_str = json.dumps(obs)
            messages.append({"role": "tool", "tool_call_id": tc.id, "name": t_name, "content": obs_str})

            steps.append({
                "step": step_idx + 1,
                "tool": t_name,
                "arguments": t_args,
                "observation": obs
            })

            # Save checkpoint
            os.makedirs("state", exist_ok=True)
            ckpt = {
                "run_id": run_id,
                "scenario_id": scenario_id,
                "messages": messages,
                "tokens_used": tokens_used,
                "step_number": step_idx + 1,
                "executed_tools": executed_tools
            }
            with open(os.path.join("state", f"{run_id}.json"), "w", encoding="utf-8") as ckpt_f:
                json.dump(ckpt, ckpt_f)

            if crash_after_tools > 0 and len(executed_tools) == crash_after_tools:
                return {"status": "CRASHED", "run_id": run_id, "executed_tools": executed_tools}

    return {
        "scenario_id": scenario_id,
        "status": "FALLBACK_MAX_ITERATIONS",
        "final_answer": "FALLBACK_MAX_ITERATIONS: exceeded max steps.",
        "llm_calls": max_iterations,
        "tokens_used": tokens_used,
        "steps": steps,
        "executed_tools": executed_tools
    }

def resume_agent(run_id):
    ckpt_path = os.path.join("state", f"{run_id}.json")
    with open(ckpt_path, "r", encoding="utf-8") as f:
        ckpt = json.load(f)
    # Resume agent loop using existing messages
    # In a full run continuation, we pass the saved messages back into chat
    return run_agent(ckpt["messages"][-1]["content"], ckpt["scenario_id"], run_id=run_id)

def execute_scenarios():
    os.makedirs("outputs", exist_ok=True)
    scenarios = pd.read_csv("data/raw/agent_scenarios.csv")
    runs = []
    for _, row in scenarios.iterrows():
        reset()
        res = run_agent(row["message"], row["scenario_id"])
        runs.append({
            "scenario_id": res["scenario_id"],
            "status": res["status"],
            "final_answer": res["final_answer"],
            "llm_calls": res["llm_calls"],
            "tokens_used": res["tokens_used"],
            "steps": res["steps"]
        })
        with open("outputs/agent_runs.jsonl", "a", encoding="utf-8") as f:
            f.write(json.dumps(runs[-1]) + "\n")
    print("Part D scenarios executed successfully.")

if __name__ == "__main__":
    execute_scenarios()
```
